```python
for i in cols:
    fig = mp.figure(figsize=(5,5))
    mp.hist(df[i],color="blue",bins=15)
    mp.xlabel(i)
    mp.show()
```

```python
years = df.columns[1:]
total_population = df[years].sum()
mp.figure(figsize=(20,20))
mp.barh( years, total_population, color="#31FFFC")
mp.xlabel("total_population",size=20)
mp.ylabel("year",size=20)
mp.title("Total Population per year",size=30)
mp.show()
```
